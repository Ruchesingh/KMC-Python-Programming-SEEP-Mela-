```python
#DataFrame with two-dimensional data with row and columns.
data={
    'Name':['Anita','Siya','Rekha'],
    'Age':[23,45,78],
    'City':['New York','KTM','MnR']
}
```

```python
# filtering dataframe
filtered_df=df[df['Age'] > 28]
filtered_df
```

```python
#multiple conditions
# And Conditions
filtered_df=df[(df['Age'] > 26) & (df['City'] == 'KTM')]
filtered_df
```

```python
# OR Conditions
filtered_df=df[(df['Age'] > 26) | (df['City'] == 'KTM')]
filtered_df
```

```python
#CRUD 
#Adding a new column
df['Salary']=[50000,30000,67000]
df
```

```python
#updating existing column
df['Salary']=df['Salary']+5000
df
```

```python
#updating existing column with conditions
df.loc[df['Age'] > 30, 'Salary']=df['Salary']+100
df
```

```python
#deleting  a column
df.drop('Salary', axis=1, inplace=True)
df
```

```python
#deleting  a row
df.drop(0, axis=0, inplace=True)
df
```
