```python
#line plot
import matplotlib.pyplot as plt
```

```python
days=['Monday','Tuesday','Wednesday','Thursday','Friday']
temperatures=[22,45,25,67,48]
plt.plot(days,temperatures,maker='o', label='Temperature')
plt.title('Temperature Over a Week')
plt.xlabel('Days')
plt.ylabel('Temperature(C)')
plt.legend()
plt.show()         
```

```python
#Line plot
days=['Monday','Tuesday','Wednesday','Thursday','Friday']
temperatures=[22,45,25,67,48]
plt.plot(days,temperatures,marker='o', label='Temperature')
plt.title('Temperature Over a Week')
plt.xlabel('Days')
plt.ylabel('Temperature(degree C)')
plt.legend()
plt.show()         
```

```python
# Bar Chart
products=['Mobile','Laptop','Tablet']
sales=[150,200,100]
plt.bar(products,sales, color=['blue','red','purple'])
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```
